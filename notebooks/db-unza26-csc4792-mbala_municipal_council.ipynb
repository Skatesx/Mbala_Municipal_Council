## 17. Exploratory Data Analysis & Visualisation

This section loads the **cleaned** dataset and examines record composition, dates, topics, and text characteristics, producing figures saved to `figures/`.

```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

FIG_DIR = Path("../figures")
FIG_DIR.mkdir(exist_ok=True)

df = pd.read_csv("../data/processed/db-unza26-csc4792-mbala_municipal_council_cleaned.csv", sep="|")
print(df.shape)
df.head()
```

### 17.1 Record type classification

**Note:** `record_type` was not carried over into the final cleaned export in the earlier cleaning step, so it is recreated here from `page_id` / `post_id` for analysis.

```python
df["record_type"] = df.apply(
    lambda r: "page" if pd.notna(r["page_id"])
    else ("post" if pd.notna(r["post_id"]) else "unclassified"),
    axis=1
)
df["record_type"].value_counts()
```

```python
counts = df["record_type"].value_counts()
plt.figure(figsize=(5,4))
counts.plot(kind="bar", color=["#4C72B0","#DD8452","#999999"])
plt.title("Distribution of Records by Type")
plt.xlabel("Record type")
plt.ylabel("Count")
plt.xticks(rotation=0)
plt.tight_layout()
plt.savefig(FIG_DIR / "record_type_distribution.png")
plt.show()
```

**Interpretation:** Of the 42 records, 22 (52%) are WordPress *posts* (news/announcements, identified by `post_id`) and 18 (43%) are static *pages* (department/CDF/finance info, identified by `page_id`). 2 records (5%) could not be classified because both IDs were missing in the scraped source — these are worth a manual check rather than discarding, since they may be valid pages with an extraction gap.

### 17.2 Date analysis

Dates are parsed with `errors="coerce"`. A meaningful share of the raw `date` values are not real dates at all (e.g. `"24 and 2025"`, `"25 November 1918"`) — likely scraping artifacts from surrounding page text — so any time-based conclusions here only cover the subset with valid timestamps and should be reported as such.

```python
df["date_parsed"] = pd.to_datetime(df["date"], errors="coerce", utc=True)

n_missing = df["date"].isna().sum()
n_invalid = df["date"].notna().sum() - df["date_parsed"].notna().sum()
n_valid = df["date_parsed"].notna().sum()
print(f"Valid dates: {n_valid} | Non-null but unparseable: {n_invalid} | Fully missing: {n_missing}")
df[df["date_parsed"].isna() & df["date"].notna()][["title","date"]]
```

```python
dated = df.dropna(subset=["date_parsed"]).copy()
dated["month"] = dated["date_parsed"].dt.tz_localize(None).dt.to_period("M").astype(str)
monthly = dated.groupby("month").size()

plt.figure(figsize=(7,4))
monthly.plot(kind="bar", color="#C44E52")
plt.title(f"Records Published Over Time (n={len(dated)} of {len(df)} with valid dates)")
plt.xlabel("Month")
plt.ylabel("Number of records")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.savefig(FIG_DIR / "records_over_time.png")
plt.show()
```

**Interpretation:** Only 22 of 42 records (52%) carry a usable date. Among those, the earliest is from October 2023 and the most recent from September 2026, with dated posts clustering in 2026 — consistent with the council's website being more actively updated with news/CDF-project posts recently than in prior years. Because roughly a third of the non-null dates were unparseable junk, this timeline should be read as indicative, not complete, and flagged as a data-quality limitation in the paper.

### 17.3 Topics / categories analysis

`topics` is a multi-label field (semicolon-separated), so it is exploded before counting.

```python
topics_exploded = df["topics"].str.split(";").explode().str.strip()
topic_counts = topics_exploded.value_counts()
topic_counts
```

```python
plt.figure(figsize=(7,4))
topic_counts.sort_values().plot(kind="barh", color="#55A868")
plt.title("Frequency of Topics/Categories Across Records")
plt.xlabel("Number of records")
plt.tight_layout()
plt.savefig(FIG_DIR / "topic_frequency.png")
plt.show()
```

**Interpretation:** `cdf` is by far the most common tag (25 of 42 records reference Constituency Development Fund content), followed by `budget_finance` (17) and `department` (9). This lines up with the assignment brief's emphasis on CDF allocations and budgets, and confirms the council's site is a reasonable source for that theme — but it also means less-represented themes (`governance`, `notice`, `development_plan`) have very few records (2–3 each), so any conclusions there should be treated cautiously given the small sample.

```python
df["is_cdf"] = topics_exploded.groupby(level=0).apply(lambda s: "cdf" in s.values)
plt.figure(figsize=(4,4))
df["is_cdf"].value_counts().rename({True:"CDF-related", False:"Not CDF-related"}).plot(
    kind="pie", autopct="%1.0f%%", colors=["#4C72B0","#DDDDDD"], ylabel=""
)
plt.title("Share of Records Tagged 'CDF'")
plt.tight_layout()
plt.savefig(FIG_DIR / "cdf_share.png")
plt.show()
```

### 17.4 Text length by record type

Comparing `text_length_chars` across `record_type` to see whether posts (news items) and pages (static info) differ in how much content they carry.

```python
df.boxplot(column="text_length_chars", by="record_type", figsize=(6,4))
plt.title("Text Length by Record Type")
plt.suptitle("")
plt.xlabel("Record type")
plt.ylabel("Text length (characters)")
plt.tight_layout()
plt.savefig(FIG_DIR / "text_length_by_record_type.png")
plt.show()

df.groupby("record_type")["text_length_chars"].describe()
```

**Interpretation:** Pages tend to run longer than posts on median text length, which makes sense — pages like the Finance department page describe ongoing structures and responsibilities, while posts are shorter, dated news announcements about specific projects. The single `unclassified` record has too small a sample (n=2) to compare meaningfully.
